```python
# Question 3: Reverse a List
numbers = [10, 20, 30, 40, 50]
l = 0
r = len(numbers)-1
while l < r:
    numbers[l], numbers[r] = numbers[r], numbers[l]
    l += 1
    r -= 1
print(numbers)
```

```python
# 17. Swap keys and values, handling duplicate values
data = {
    "Alice": "IT",
    "Bob": "HR",
    "Charlie": "IT"
}

d = {}

for i, j in data.items():
    if j in d:
        d[j].append(i)
    else:
        d[j] = [i]

print(d)
```

```python
# 18. Sort a dictionary by its values
data = {
    "Alice": 80,
    "Bob": 95,
    "Charlie": 70
}

result = {}

while data:
    mini = float('inf')
    name = None

    for i, j in data.items():
        if j < mini:
            mini = j
            name = i

    result[name] = mini
    del data[name]

print(result)
        
```

```python
# 20. Nested Dictionary — Calculate Total Marks
students = {
    "Alice": {
        "math": 80,
        "science": 90
    },
    "Bob": {
        "math": 70,
        "science": 85
    }
}

total = 0
for name, marks in students.items():
    for subject, mark in marks.items():
        total += mark
    print(name, "→", total)
```
